```python
############# WRITE YOUR CODE IN THIS CELL (IF APPLICABLE)  ####################


# Import required libraries
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.metrics import classification_report, confusion_matrix

# Load the IMDB dataset
# Only the top 10,000 most frequent words are kept
num_words = 10000
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=num_words)

# Pad all sequences to the same length
# This is required because LSTM expects equal-length input sequences
max_length = 200
X_train = pad_sequences(X_train, maxlen=max_length)
X_test = pad_sequences(X_test, maxlen=max_length)

# Build the LSTM model
model = Sequential()

# Embedding layer converts word indexes into dense vectors
model.add(Embedding(input_dim=num_words, output_dim=128, input_length=max_length))

# LSTM layer learns patterns in sequential data
model.add(LSTM(128))

# Dropout helps reduce overfitting
model.add(Dropout(0.5))

# Output layer for binary classification (positive / negative)
model.add(Dense(1, activation="sigmoid"))

# Compile the model
# Binary cross-entropy is used for binary sentiment classification
model.compile(
    loss="binary_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# Train the LSTM model
model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)

# Evaluate the model on test data
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)

# Generate predictions
y_pred = (model.predict(X_test) > 0.5).astype("int32")

# Display detailed evaluation metrics
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\ncompleted successfully")
```

**Sentiment Classification Using an LSTM Model on the IMDB Dataset**

*Introduction*

In this task, a **Long Short-Term Memory (LSTM)** neural network was implemented to perform **sentiment classification** on a sequential text dataset. The **IMDB movie reviews dataset** was selected as it is a widely used benchmark dataset for natural language processing tasks. The objective was to classify movie reviews as either **positive or negative**, demonstrating the ability of recurrent neural networks to learn long-term dependencies in sequential data.

*Dataset Description and Preprocessing*

The IMDB dataset contains **50,000 movie reviews**, evenly split into training and testing sets. Each review is represented as a sequence of word indices corresponding to the most frequent words in the dataset. In this implementation, only the **top 10,000 most frequent words** were retained to reduce vocabulary size and computational complexity.

Since LSTM networks require input sequences of equal length, all reviews were padded to a fixed length of **200 tokens** using post-padding. This ensures consistent input dimensions while preserving the most relevant parts of each review.

*Model Architecture*

The model was constructed using a **Sequential** architecture consisting of the following layers:

* An **Embedding layer**, which converts word indices into dense 128-dimensional vector representations. This allows the model to learn semantic relationships between words.
* An **LSTM layer with 128 units**, which captures temporal dependencies and contextual information across word sequences.
* A **Dropout layer (0.5)**, introduced to reduce overfitting by randomly deactivating neurons during training.
* A **Dense output layer** with a sigmoid activation function, suitable for binary classification tasks.

*Model Training*

The model was trained using the **Adam optimizer** and **binary cross-entropy loss**, which is appropriate for binary sentiment classification. Training was performed for **5 epochs** with a batch size of **64**, using **20% of the training data for validation**. During training, both training and validation accuracy improved consistently, indicating effective learning without significant overfitting.

*Evaluation and Performance Metrics*

After training, the model was evaluated on the unseen test dataset. The evaluation included:

* **Test accuracy**, which reflects the overall classification performance
* A **classification report**, providing precision, recall, and F1-score for both sentiment classes
* A **confusion matrix**, offering insight into correct and incorrect predictions

The results demonstrate that the LSTM model achieved strong predictive performance, successfully distinguishing between positive and negative reviews. The confusion matrix shows a balanced distribution of true positives and true negatives, confirming the robustness of the model.

*Conclusion*

This experiment highlights the effectiveness of **LSTM networks for sequential data modeling**, particularly in natural language processing tasks such as sentiment analysis. By leveraging word embeddings and memory-based recurrent structures, the model was able to capture contextual information across long text sequences. The achieved accuracy and evaluation metrics confirm that LSTM-based models are well-suited for text classification problems and can significantly outperform traditional machine learning approaches that lack sequential awareness.
