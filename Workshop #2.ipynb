```python
# exercise 2 solution
def encode_sex(data):
    data.reset_index(drop=True, inplace=True) # reindexing the dataset in case the dataset index is corrupted
    rows=data.shape[0]
    a=data.loc[:,'sex']
    for i in range(rows): 
        if a[i]=="Male":
            data.loc[i,"sex"]="M"
        elif a[i]=="Female":
            data.loc[i,"sex"]="F"
    return data
```

```python
                                    ############# WRITE YOUR CODE IN THIS CELL (IF APPLICABLE)  ####################
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# ---------- STEP 1: Define numerical columns explicitly ----------
numeric_columns = [
    "age", "fnlwgt", "education-num",
    "capital-gain", "capital-loss", "hours-per-week"
]

# convert numerical columns to numeric type
data_10k[numeric_columns] = data_10k[numeric_columns].apply(
    pd.to_numeric, errors="coerce"
)

# drop rows with newly created NaN values
data_10k = data_10k.dropna().reset_index(drop=True)

# ---------- STEP 2: Identify categorical & numerical columns ----------
categorical_cols = data_10k.select_dtypes(include=["object"]).columns
numerical_cols = data_10k.select_dtypes(include=["int64", "float64"]).columns

# ---------- STEP 3: Normalize numerical columns ----------
scaler = MinMaxScaler()
data_10k[numerical_cols] = scaler.fit_transform(data_10k[numerical_cols])

# ---------- STEP 4: Encode categorical columns ----------
data_encoded = pd.get_dummies(data_10k, columns=categorical_cols)

# ---------- STEP 5: Print final dataset ----------
print(data_encoded)

```

**P 2.3: Write a function to split the dataset in half column-wise and swap the first half and the second half. Apply the funtion to the dataset and print the result (3%)**

```python
                                ############# WRITE YOUR CODE IN THIS CELL (IF APPLICABLE)  ####################
def swap_column_halves(data):
    # number of columns
    n_cols = data.shape[1]
    mid = n_cols // 2

    # split and swap
    first_half = data.iloc[:, :mid]
    second_half = data.iloc[:, mid:]

    swapped_data = pd.concat([second_half, first_half], axis=1)
    return swapped_data


# apply function to dataset
swapped_dataset = swap_column_halves(data_encoded)

# print result
print(swapped_dataset)
```

After splitting the dataset column-wise, the positions of the two halves were swapped so that the second half of the columns appears first, followed by the first half. This transformation was performed without altering the data values, ensuring that only the column order was changed.

The function was applied to the processed dataset, and the resulting dataset with swapped column halves was printed. This operation demonstrates effective use of Pandas indexing and concatenation for advanced data manipulation tasks.
