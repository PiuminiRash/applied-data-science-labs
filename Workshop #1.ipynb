```python
# Example data
x = [1, 2, 3, 4]  # Predictor
y = [2.2, 2.8, 4.5, 3.7]  # Response
```

```python
########################################################This is a code cell for Task 1####################################################

import matplotlib.pyplot as plt
from scipy.stats import linregress
import numpy as np

# Given data
x = [1, 2, 3, 4]   # Predictor
y = [2.2, 2.8, 4.5, 3.7]   # Response

# (a) Plot y against x
plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Scatter Plot of y vs x")
plt.show()

# (b) Run simple linear regression
result = linregress(x, y)
m = result.slope
b = result.intercept

print("Slope (m):", m)
print("Intercept (b):", b)

# (c) Predict y values
y_pred = [m * xi + b for xi in x]
print("Predicted y values:", y_pred)

# (d) Plot original and predicted values
plt.scatter(x, y, label="Original y-values")
plt.plot(x, y_pred, label="Predicted y-values")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Original vs Predicted y-values")
plt.legend()
plt.show()

# (e) Accuracy measurement
r_squared = result.rvalue ** 2
mse = np.mean((np.array(y) - np.array(y_pred)) ** 2)

print("R-squared:", r_squared)
print("Mean Squared Error:", mse)
```

                                                    Report Cell – Task 1                                                               
*Explanation*<br>
A scatter plot was used to examine the relationship between the predictor variable **x** and the response variable **y**.  
The `linregress` function from the **SciPy** library was used to build a simple linear regression model.

The regression equation is:
\[
y = mx + b
\]

Where:  
- **m** represents the slope  
- **b** represents the intercept  

Predicted y-values were calculated using the regression equation.  
A comparison graph was plotted to visually assess the difference between the original y-values and the predicted y-values.

*Accuracy Evaluation*<br>

Model accuracy was measured using the following metrics:

- **R² (R-squared):** Indicates how well the independent variable *x* explains the variance in *y*  
- **Mean Squared Error (MSE):** Measures the average squared difference between actual and predicted values  

A **higher R²** value and a **lower MSE** indicate better model performance.

Repeat task 1 for the following set of points X and y.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Generate example data for multiple regression
np.random.seed(42)
X1 = np.random.rand(100) * 10  # Predictor 1
X2 = np.random.rand(100) * 20  # Predictor 2


# Combine predictors into a single matrix
X = np.column_stack((X1, X2))

# Response variable with noise
y = 4 + 3 * X1 + 2 * X2 + np.random.randn(100) * 5
```

```python
###############3# This is a code cell for Task 2#######

import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from mpl_toolkits.mplot3d import Axes3D

# Generate example data
np.random.seed(42)
X1 = np.random.rand(100) * 10   # Predictor 1
X2 = np.random.rand(100) * 20   # Predictor 2

# Combine predictors into a single matrix
X = np.column_stack((X1, X2))

# Response variable with noise
y = 4 + 3 * X1 + 2 * X2 + np.random.randn(100) * 5

# Fit multiple linear regression model
model = LinearRegression()
model.fit(X, y)

# Get coefficients
m1, m2 = model.coef_
b = model.intercept_

print("Coefficient m1:", m1)
print("Coefficient m2:", m2)
print("Intercept b:", b)

# Predict y values
y_pred = model.predict(X)

# Measure accuracy
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Squared Error:", mse)
print("R-squared:", r2)

# 3D plot of original and predicted values
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(X1, X2, y, label="Original y", alpha=0.6)
ax.scatter(X1, X2, y_pred, label="Predicted y", alpha=0.6)

ax.set_xlabel("X1")
ax.set_ylabel("X2")
ax.set_zlabel("y")
ax.set_title("Multiple Linear Regression: Original vs Predicted Values")
ax.legend()

plt.show()
```

                                                        Report Cell – Task 2                                                               
*Explanation*<br>
This task extends Task 1 to **multiple linear regression**, where the response variable **y** depends on two predictor variables (**X1** and **X2**).

The multiple linear regression model is defined as:<br>
\[
y = b + m_1X_1 + m_2X_2
\]

Where:  
- **m₁, m₂** are coefficients for the predictors  
- **b** is the intercept  

The `LinearRegression` class from the **scikit-learn** library was used to train the model.

*Prediction and Visualization*<br>
Predicted y-values were generated using the trained model.  
A **3D scatter plot** was used to visualize:
- Original y-values  
- Predicted y-values  

This helps to visually assess how well the regression plane fits the data.

*Accuracy Evaluation*<br>
Model performance was evaluated using:
- **Mean Squared Error (MSE):** Measures prediction error  
- **R² (R-squared):** Indicates how well the predictors explain the variance in y  

A **higher R²** and **lower MSE** indicate better predictive performance.

*Conclusion*<br>
The multiple linear regression model successfully captures the relationship between the predictors and the response variable, despite the presence of noise in the data.
