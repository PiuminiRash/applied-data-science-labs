# --------- Predicting employees' productivity using regression models ---------

```python
import pandas as pd
import numpy as np
```

```python
# See https://archive.ics.uci.edu/dataset/597/productivity+prediction+of+garment+employees for the dataset description
# the goal is to predict employees' productivity 
# replace the file path with the file path on your local disk
data=pd.read_csv('csv/garments_worker_productivity.csv')
```

```python
data
```

```python
data.shape
```

```python
#checking null values
data.isna().sum()
```

```python
#deleting null values
data.dropna(inplace=True)
```

```python
# deleting the date column
data=data.drop('date',axis=1)
```

```python
# extrcting input and output data
X=data.drop('actual_productivity',axis=1)
y=data['actual_productivity']
```

```python
# Encoding categorical columns and then normalising the input data
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder

ct = ColumnTransformer(
 [("t1", OrdinalEncoder(), [0,2]), ("t2", OneHotEncoder(drop="first"), [1])],remainder='passthrough')

X=ct.fit_transform(X)
scaler=MinMaxScaler()
X=scaler.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
# Training the models
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Ridge
reg = LinearRegression().fit(X_train,y_train)
ridge = Ridge(alpha=0.3).fit(X_train,y_train)
```

```python
# Predicting
Pred_reg=reg.predict(X_test)
Pred_ridge=ridge.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

print('reg mean square error is {}'.format(mean_squared_error(y_test,Pred_reg)))
print('\nreg mean absolute error is {}'.format(mean_absolute_error(y_test,Pred_reg)))
print('\nreg R^2 score is {}'.format(r2_score(y_test,Pred_reg)))

print('\n\nridge mean square error is {}'.format(mean_squared_error(y_test,Pred_ridge)))
print('\nridge mean absolute error is {}'.format(mean_absolute_error(y_test,Pred_ridge)))
print('\nridge R^2 score is {}'.format(r2_score(y_test,Pred_ridge)))
```
