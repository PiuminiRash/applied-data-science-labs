```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Sample data
x = np.array([1, 2, 3, 4, 5])  # Independent variable
y = np.array([2, 4, 5, 4, 5])  # Dependent variable

# (a) Plot y against x
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', label='Original Data')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Scatter Plot of y vs. x')
plt.legend()
plt.show()

# (b) Run a simple regression model
x_reshaped = x.reshape(-1, 1)  # Reshape x for sklearn
model = LinearRegression()
model.fit(x_reshaped, y)
m = model.coef_[0]  # Slope
b = model.intercept_  # Intercept

print(f"Coefficient (m): {m}")
print(f"Intercept (b): {b}")

# (c) Predict y values using the regression model
y_pred = model.predict(x_reshaped)
print(f"Predicted y-values: {y_pred}")

# (d) Plot two graphs on the same plane
plt.figure(figsize=(8, 5))
plt.plot(x, y_pred, color='red', label='Predicted y-values')
plt.scatter(x, y, color='blue', label='Original y-values')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Comparison of Predicted and Original y-values')
plt.legend()
plt.show()

# (e) Measure accuracy
mse = mean_squared_error(y, y_pred)  # Mean Squared Error
r2 = r2_score(y, y_pred)  # R-squared value

print(f"Mean Squared Error (MSE): {mse}")
print(f"R-squared (R2) Value: {r2}")

# Accuracy Analysis
if r2 > 0.8:
    print("The model provides a good fit to the data.")
elif r2 > 0.5:
    print("The model provides a moderate fit to the data.")
else:
    print("The model does not fit the data well.")


# Task 2: Multiple regression
# Generate example data
np.random.seed(42)
X1 = np.random.rand(100) * 10  # Predictor 1
X2 = np.random.rand(100) * 20  # Predictor 2
X = np.column_stack((X1, X2))  # Combine predictors into a matrix
true_b = 4
true_m1, true_m2 = 3, 2

# Response variable with noise
y_multi = true_b + true_m1 * X1 + true_m2 * X2 + np.random.randn(100) * 5

# Fit multiple regression model
multi_model = LinearRegression()
multi_model.fit(X, y_multi)
m1, m2 = multi_model.coef_
b_multi = multi_model.intercept_
print(f"Multiple Regression Coefficients: m1 = {m1}, m2 = {m2}, b = {b_multi}")

# Predict y values
y_multi_pred = multi_model.predict(X)

# Plot comparison for multiple regression (original vs predicted)
def plot_multi_comparison():
    plt.figure(figsize=(10, 5))
    plt.scatter(range(len(y_multi)), y_multi, color='blue', label='Original y')
    plt.plot(range(len(y_multi)), y_multi_pred, color='red', label='Predicted y')
    plt.xlabel('Index')
    plt.ylabel('y')
    plt.title('Original y vs. Predicted y for Multiple Regression')
    plt.legend()
    plt.show()

plot_multi_comparison()

# Measure accuracy for multiple regression
mse_multi = mean_squared_error(y_multi, y_multi_pred)
r2_multi = r2_score(y_multi, y_multi_pred)
print(f"Multiple Regression Mean Squared Error: {mse_multi:.2f}")
print(f"Multiple Regression R-squared: {r2_multi:.2f}")
```
